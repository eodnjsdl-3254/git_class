## 필수 1. 리스트와 튜플의 가변성 비교하기

### 문제 1-1. 주문 목록 수정과 고정 좌표 확인

변경 가능한 주문 목록은 리스트로, 변경되면 안 되는 위치 좌표는 튜플로 관리합니다.

### 요구사항

1. `"Americano"`, `"Cafe Latte"`, `"Cheese Cake"`가 포함된 `order_items` 리스트를 생성하세요.
2. 두 번째 요소 `"Cafe Latte"`를 `"Vanilla Latte"`로 수정하세요.
3. 수정된 전체 리스트를 출력하세요.
4. 서울의 위도와 경도를 의미하는 `(37.5665, 126.9780)`을 `office_coordinate` 튜플에 저장하세요.
5. 첫 번째 값인 위도를 출력하세요.
6. 마지막 값인 경도를 출력하세요.
7. 아래 코드를 직접 실행하지 말고 주석으로 작성하여 튜플 수정 시 오류가 발생함을 확인하세요.

```python
# office_coordinate[0] = 38.0
```

### 예상 핵심 결과

```text
['Americano', 'Vanilla Latte', 'Cheese Cake']
37.5665
126.978
```

````python
# 문제 1-1 코드를 작성하세요.

# 1. 주문 목록 리스트 생성 (가변 자료형)
order_items = ["Americano", "Cafe Latte", "Cheese Cake"]  #

# 2. 두 번째 요소(인덱스 1)를 "Vanilla Latte"로 수정
order_items[1] = "Vanilla Latte"  #

# 3. 수정된 전체 리스트 출력
print(order_items)  #

# 4. 서울의 위도와 경도를 튜플에 저장 (불변 자료형)
office_coordinate = (37.5665, 126.9780)  #

# 5. 첫 번째 값(위도) 정방향 인덱싱으로 출력
print(office_coordinate[0])  #

# 6. 마지막 값(경도) 역방향 인덱싱으로 출력
print(office_coordinate[-1])  #

# 7. 튜플은 불변(Immutable) 자료형이므로 요소를 수정하려고 하면 TypeError가 발생함
# office_coordinate[0] = 38.0
````

### 확인 사항

- 리스트는 요소 수정이 가능한지 확인하세요.
- 튜플은 요소 수정이 불가능한 불변 자료형이라는 점을 확인하세요.
- 첫 번째 요소는 `0`, 마지막 요소는 `-1` 인덱스로 추출하세요.

## 필수 2. 인덱싱과 슬라이싱으로 데이터 구간 추출하기

### 문제 2-1. 시간대별 주문 건수 일부 추출

0시부터 5시까지의 주문 건수가 순서대로 저장되어 있다고 가정합니다.

### 요구사항

1. `[12, 15, 9, 20, 18, 11]`을 `hourly_orders` 리스트에 저장하세요.
2. 첫 번째 주문 건수를 인덱싱으로 출력하세요.
3. 마지막 주문 건수를 인덱싱으로 출력하세요.
4. 인덱스 1부터 3까지의 값을 슬라이싱으로 출력하세요.
5. 처음부터 끝까지 두 칸 간격으로 추출하세요.
6. 전체 데이터를 역순으로 출력하세요.

### 예상 출력

```text
12
11
[15, 9, 20]
[12, 9, 18]
[11, 18, 20, 9, 15, 12]
```

````python
# 문제 2-1 코드를 작성하세요.

# 1. 시간대별 주문 건수 리스트 저장
hourly_orders = [12, 15, 9, 20, 18, 11]  #[cite: 1]

# 2. 첫 번째 주문 건수 출력 (정방향 인덱스 0)
print(hourly_orders[0])  #[cite: 1]

# 3. 마지막 주문 건수 출력 (역방향 인덱스 -1)
print(hourly_orders[-1])  #[cite: 1]

# 4. 인덱스 1부터 3까지 슬라이싱 (끝 인덱스 4는 포함되지 않음)
print(hourly_orders[1:4])  #[cite: 1]

# 5. 처음부터 끝까지 두 칸 간격(증감폭 2)으로 추출
print(hourly_orders[::2])  #[cite: 1]

# 6. 전체 데이터를 역순(증감폭 -1)으로 출력
print(hourly_orders[::-1])  #[cite: 1]
````

### 확인 사항

- `[1:4]`에서 끝 인덱스 4는 포함되지 않습니다.
- `[::2]`는 두 칸 간격 추출입니다.
- `[::-1]`은 원본을 변경하지 않고 역순의 새 리스트를 만듭니다.

## 필수 3. 시퀀스 공통 연산과 권한 확인하기

### 문제 3-1. 사용자 그룹 병합과 권한 검증

### 요구사항

1. `[101, 102]`를 `admin_group1`에 저장하세요.
2. `[103, 104]`를 `admin_group2`에 저장하세요.
3. `+` 연산자로 두 그룹을 결합하여 `all_admins`에 저장하고 출력하세요.
4. `admin_group1 * 3`의 결과를 출력하세요.
5. `"admin"`, `"editor"`, `"member"`가 포함된 `user_roles` 리스트를 생성하세요.
6. `len()`으로 역할 개수를 출력하세요.
7. `"admin" in user_roles`의 결과를 출력하세요.
8. `"guest" not in user_roles`의 결과를 출력하세요.

### 예상 출력

```text
[101, 102, 103, 104]
[101, 102, 101, 102, 101, 102]
3
True
True
```

````python
# 문제 3-1 코드를 작성하세요.

# 1 & 2. 두 관리자 그룹 리스트 생성
admin_group1 = [101, 102]  #[cite: 1]
admin_group2 = [103, 104]  #[cite: 1]

# 3. '+' 연산자로 두 리스트를 결합하여 새로운 리스트 생성 및 출력
all_admins = admin_group1 + admin_group2  #[cite: 1]
print(all_admins)  #[cite: 1]

# 4. '*' 연산자로 리스트 요소를 3번 반복한 결과 출력
print(admin_group1 * 3)  #[cite: 1]

# 5. 사용자 역할 리스트 생성
user_roles = ["admin", "editor", "member"]  #[cite: 1]

# 6. len() 함수로 시퀀스의 길이(요소 개수) 출력
print(len(user_roles))  #[cite: 1]

# 7. 'in' 연산자로 특정 요소 포함 여부 확인 (True 반환)
print("admin" in user_roles)  #[cite: 1]

# 8. 'not in' 연산자로 특정 요소 미포함 여부 확인 (True 반환)
print("guest" not in user_roles)  #[cite: 1]
````

### 확인 사항

- `+`와 `*` 연산은 원본을 직접 수정하지 않고 새로운 시퀀스를 만듭니다.
- 서로 다른 자료형인 리스트와 튜플은 `+`로 바로 결합할 수 없습니다.
- `in`, `not in`의 결과는 불리언 값입니다.

## 심화 1. 패킹과 일반 언패킹

### 문제 4-1. 주문 요약 정보 묶기와 분리하기

하나의 주문 정보를 주문번호, 상품명, 결제금액 순서로 묶고 다시 분리합니다.

### 요구사항

1. 다음 세 값을 쉼표로 나열하여 `packed_order`에 저장하세요.
   - 주문번호: `1001`
   - 상품명: `"Americano"`
   - 결제금액: `4500`
2. `packed_order`와 자료형을 출력하세요.
3. `order_id`, `menu_name`, `payment_amount` 세 변수로 일반 언패킹하세요.
4. 분리된 세 값을 각각 출력하세요.

### 예상 핵심 결과

```text
(1001, 'Americano', 4500)
<class 'tuple'>
1001
Americano
4500
```

````python
# 문제 4-1 코드를 작성하세요.

# 1. 괄호 없이 쉼표로 값을 나열하여 튜플로 패킹(Packing)
packed_order = 1001, "Americano", 4500  #[cite: 1]

# 2. 패킹된 변수 값과 자료형(<class 'tuple'>) 출력
print(packed_order)  #[cite: 1]
print(type(packed_order))  #[cite: 1]

# 3. 요소 개수와 동일한 수의 변수로 일반 언패킹(Unpacking)
order_id, menu_name, payment_amount = packed_order  #[cite: 1]

# 4. 분리된 세 값을 각각 출력
print(order_id)  #[cite: 1]
print(menu_name)  #[cite: 1]
print(payment_amount)  #[cite: 1]
````

### 확인 사항

- 괄호를 쓰지 않아도 쉼표로 나열한 값이 튜플로 패킹되는지 확인하세요.
- 요소 수와 변수 수가 같아야 일반 언패킹이 가능합니다.
